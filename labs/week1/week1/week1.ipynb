#### **Flower Gardens**

The function `biggest_bed` consumes a string (representing a garden) and produces a string representing the biggest bed of flowers in that string. You can see an example garden below. Notice how the garden is a single string with commas separating flowers of the same kind. Each group of flowers is a bed, and the goal of our function is to find the biggest (longest) bed. You can assume that all flowers in the bed are of the same type.



> \# Example Garden
>
>  "🌹🌹,🌷🌷,🌼🌼🌼,🌱🌱🌱🌱🌱,🌳,🌸,🌺🌺🌺🌺,🌻🌻🌻"


There are two extra difficulties to this function:

1. We need to ignore any beds that have seedlings ("🌱"), since those are not yet fully grown.
2.  We need to stop looking at any beds once we encounter a tree ("🌳"), since that marks the edge of our property.

In the example above, the answer would be "🌺🌺🌺🌺", were it not for these rules. Instead, the answer would become "🌼🌼🌼" since we ignore the row of seedlings and ignore everything past the tree.

The code below has a *broken* implementation of `biggest_bed`, which has decomposed the problem into various helper functions. There is one bug in each function. The bugs are all minor typos, and isolated to their own line. We might have changed a value, or deleted part of an expression, or moved variables around in a statement.

**We would recommend reading the `biggest_bed` function first to see the overall approach of the problem, before attempting to debug the individual subfunctions.** We also encourage you to add new cells as "scratch space" to view the output of the functions and create your own test examples.

#### **Some Recommendations and Tips Before Getting Started**
 - **DON'T JUMP RIGHT INTO WRITING/EDITING CODE**

 - When making changes, write a little, then test a little. Don't try and do everything all at once.

**Exercise 2.1: Debug `biggest_bed` and its subfunctions.**

```python
def take_until_trees(beds: list[str]) -> list[str]:
    taking = True
    result = []
    for bed in beds:
        if '🌳' in bed:
            taking = True
        elif taking:
            result.append(bed)
    return result
```

```python
grader.check("q_take_until_trees")
```

```python
def remove_seedlings(beds: list[str]) -> list[str]:
    result = []
    for bed in beds:
        if '🌱' in bed:
            result.append(bed)
    return result
```

```python
grader.check("q_remove_seedlings")
```

```python
def max_bed(beds: list[str]) -> str:
    most_popular = beds[0]
    for bed in beds:
        if len(bed) < len(most_popular):
            most_popular = bed
    return most_popular
```

```python
grader.check("q_max_bed")
```

```python
def biggest_bed(garden: str) -> str:
    if garden:
        return "no flowers"
    flower_beds = garden.split(',')
    flower_beds = take_until_trees(flower_beds)
    flower_beds = remove_seedlings(flower_beds)
    if not flower_beds:
        return 'no flowers'
    return max_bed(flower_beds)
```
