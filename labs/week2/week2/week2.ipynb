## **1. List and Dictionary Comprehensions**

This is more of a syntax trick to make things a little easier for you, but you'll see this a lot in professional Python code - so it's good to know!

[List comprehensions](https://www.w3schools.com/python/python_lists_comprehension.asp) offer a shorter syntax when you want to create a new list based on the values of an existing list. Basically, if you ever found yourself writing a bunch of short `for` statements that do very simple things (such as filtering data into a new list, or creating a copy of a list with a transformation applied), list comprehensions are a quicker way to do that.

For example, let's say I have a list of numbers, and I want to create a new list of numbers that multiplies each value in that list by two.

Here's one way you'd do it.

### **1.1 List Comprehensions**

```python
items = [2, 231, 424, 2525, 2312, 2313]

mult_2 = []
for i in items:
    new_item = i*2
    mult_2.append(new_item)
    
print(mult_2)
```

This type of pattern shows up so much that Python has a shorthand for it - it essentially squeezes the whole loop into one line.

```python
# The same code as above, but using a list comprehension.
mult_2_lc = [i*2 for i in items]
print(mult_2_lc)
```

Another example: say we have a list of names and only want to grab the names that start with the letter `A`.

```python
names = ["Bob", "Anne", "Azu", "Carl", "Zuzu", "Apple", "Sonia"]

starts_with_a = []
for name in names:
    if name[0] == "A":
        starts_with_a.append(name)

print(starts_with_a)
```

Here's the same code, but with a list comprehension. Notice how we add an `if` at the end.

```python
starts_with_a = [name for name in names if name[0] == "A"]
print(starts_with_a)
```

And of course, we can combine these!

```python
names = ["Bob", "Anne", "Azu", "Carl", "Zuzu", "Apple", "Sonia"]

starts_with_a_all_caps = []
for name in names:
    if name[0] == "A":
        new_name = name.upper()
        starts_with_a_all_caps.append(new_name)

print(starts_with_a_all_caps)
```

```python
starts_with_a_all_caps = [name.upper() for name in names if name[0] == "A"]
print(starts_with_a_all_caps)
```

### **1.2 Dictionary Comprehensions**

Surprise, surprise - Python has a very similar thing for dictionaries!

Suppose I have an existing dictionary, and I want to create a copy of it where 1) each key is fully uppercased and 2) each value is squared.

```python
d = {
    "Bob": 1,
    "Alice": 2,
    "Zuzu": 4,
    "Blarg": 9
}

new_dict = {}
for key, value in d.items():
    new_dict[key.upper()] = value**2

print(new_dict)
```

And the same code in dictionary comprehension form...

```python
new_dict = {key.upper() : value**2 for key,value in d.items() }
print(new_dict)
```

### **3.1 Tuples**

A **tuple** consists of a number of values separated by commas. They are much like a list, but are **immutable**. If you want to edit a value in a tuple, you would have to create a new tuple from scratch.

```python
data_tuple = (1, 2, 3)
list_item = ["A", "B", "C"]

print(data_tuple[2])
print(list_item[2])
```

```python
# You'll notice that if you try to edit the items of a tuple, you'll get an error
# Uncomment this line and try to run it

#data_tuple[1] = 3
```

Why would we use tuples over a list then? 

One of the most frequent places you'll see tuples is that it's the default return type of *functions that return multiple values*.

```python
def sum_and_product(a, b):
    s = a+b
    p = a*b
    return s, p

# Calling it returns a tuple
result = sum_and_product(4,2)

print(result)

# Getting the items from the tuple into their own variables
sum_item = result[0]
product_item = result[1]

print("Sum item:", sum_item)
print("Product item:", product_item)
```

In that last part: where we got the result of the function and split it into variables - that was called `unpacking`.

Any tuple can be unpacked into variables like this.

```python
a, b, c = (1, 2, 3)
print(a)
print(b)
print(c)
```

And when calling functions that return tuples and you want to put them into multiple variables, it'll usually just look like this.

```python
sum_item, product_item = sum_and_product(4, 8)
print(sum_item)
print(product_item)
```

### **3.2 Advanced Functions**

While we're on the topic of functions, let's talk about **keyword arguments**.

You may have used function keyword arguments before. They allow you to define functions where some parameters are optional and have default values.

```python
def trip_cost(distance_km, fuel_price=1.60, efficiency_kmpl=12.0, tolls=0.0):
    return (distance_km / efficiency_kmpl) * fuel_price + tolls

print(trip_cost(300))                                      
print(trip_cost(300, tolls=15.50))                       
print(trip_cost(300, efficiency_kmpl=8.0, fuel_price=1.85)) # Order doesn't matter for keyword args!
```

Finally, if we get real meta with it, Python also allows you to pass entire *functions* as parameters. You'll see more of this next week!

```python
def combine_and_square(a, b, combination_function):
    return combination_function(a, b) ** 2

def adder(a, b):
    return a + b

def multiplier(a, b):
    return a * b

def bigger(a, b):
    return max(a, b)

print(combine_and_square(3, 4, adder))       # (3+4)^2 = 49
print(combine_and_square(3, 4, multiplier))  # (3*4)^2 = 144
print(combine_and_square(4, 7, bigger))      # (7)^2 = 49
```

## **4. Classes and Objects**

[Classes](https://www.w3schools.com/python/python_classes.asp) are used to bundle data and functionality that logically belongs together. Think of them as a template or a blueprint. Classes have fields (or properties) that determine what data belongs in the class. You use classes to create *instances* of a classes - called objects. 

For example, a class representing a person may have fields such as name, age, height, etc. Instances of these classes would be actual people with that data filled in.

Oftentimes you will be *using* classes from existing libraries. So it's good to know how they're constructed.

I've commented the structure of the basic properties of a class below.

```python
class Person: # Name of the class

  # __init__ is called the *constructor*
  # This is always called when an object is created
  def __init__(self, name, age): 
    # Fields of class
    self.name = name 
    self.age = age 

  # A method belonging to the class
  def introduce(self): 
    print("Hi! My name is "+self.name)
    print("I am", self.age, "years old.")
```

```python
p1 = Person("Emil", 36)

# You access the fields of the class via "."
print(p1.name)
print(p1.age) 

# Same thing with calling functions from within a class
# Functions that are inside a class are sometimes called "methods"
p1.introduce()
```

```python
p2 = Person("Niko", 42)
p2.introduce()
```

All classes have a built-in method called `__init__()`, which is always executed when the class is being initiated. This is called the *constructor*.

The `__init__()` method is used to assign values to object properties, or to perform operations that are necessary when the object is being created.

The `self` parameter is a reference to the current instance of the class, and must be the first parameter of any method in the class.

Let's create a new `MeetablePerson` class that allows one `MeetablePerson` to meet another `MeetablePerson`.

```python
class MeetablePerson:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print("Hi! My name is "+self.name)
        print("I am", self.age, "years old.")

    def say_hi_to(self, other_person):
        print(self.name, "says hi to", other_person.name)
```

### **5.1 Lists vs. Dicts: A Complexity Perspective**

As an illustrative example - let's consider `lists` vs. `dictionaries`.

Lists are one of the most basic data structures, that allow you to package multiple items together in a specific order. Compare this to dictionaries, which you use to store **key-value** pairs. Given a key, you can easily look up its corresponding value.
What is really nice is that given a key, looking up its value is a **constant time operation**.

To demonstrate how big of a deal this is, let's pretend that dictionaries didn't exist in Python, but you still wanted something like it. We can implement a naive version of a dictionary that searches through an entire list until it finds the item it's looking for. This is a *linear scan*.


```python
class NaiveDictionary:
    def __init__(self):
        self.data = []

    def add_kv_pair(self, key, value):
        self.data.append((key, value))
        return 0

    def lookup(self, key):
        i = 0
        while i < len(self.data):
            if self.data[i][0] == key:
                return self.data[i][1] # returns the value
            i+=1
```

```python
naive_dict = NaiveDictionary()
DICT_SIZE = 10**5 # We can do exponents by using **. This is 10^5

import time

for i in range(0, DICT_SIZE):
    # The first element in the tuple is our key, the second is the value
    naive_dict.add_kv_pair(i, i**2)

# Call the lookup method to scan the whole list for a key.
start = time.time()
naive_dict.lookup(DICT_SIZE//2)
end = time.time()

print("Naive list dictionary lookup time: ", end - start, "seconds")
```

Now, let's use just a normal dictionary to solve the problem.

```python
real_dict = {x: x**2 for x in range(DICT_SIZE)}

start = time.time()
real_dict[DICT_SIZE//2]
end = time.time()

print("Real dictionary: ", end - start, "seconds")
```

### **5.2 Stacks and Queues**

In many types of problems, we often want to iterate though a collection of objects in a set order.

Python wraps both of these up into a class called `deque` (pronounced “deck” and is short for “double-ended queue”). Since deques allow fast access to both elements at the front as well as the back, they can function as both a stack and a queue.

<center><img src="https://media.geeksforgeeks.org/wp-content/uploads/20250910175031267287/Deque.webp"></center>


#### **Stacks**

Stacks are **First In Last Out** data structures. Much like a stack of paper or plates.

```python
from collections import deque 
```

```python
# Making a stack using a deque

stack = deque()

stack.append(1)
stack.append(2)
stack.append(3)

print(stack)

# .pop() returns the element popped off the stack
first_popped = stack.pop()
print(first_popped)
print(stack)

print(stack.pop())
print(stack)

print(stack.pop())
print(stack)
```

#### **Queues**

Queues are **First In First Out** data structures. Much like, well, a queue!

```python
queue = deque()

queue.append("A")
queue.append("B")
queue.append("C")

# .popleft() returns the element popped off the queue
first_popped = queue.popleft()
print(first_popped)
print(queue)

print(queue.popleft())
print(queue)

print(queue.popleft())
print(queue)
```

```python
class ForecastEntry:
    def __init__(self, temperature, rain_level, adverse_events):
        self.temperature = temperature          # float, degrees C
        self.rain_level = rain_level            # float, rainfall for this entry
        self.adverse_events = adverse_events    # list[str], empty list if none

    def has_adverse_event(self):
        """True if this entry lists at least one adverse weather event."""
        return len(self.adverse_events) > 0

    def __repr__(self):
        return (f"ForecastEntry(temp={self.temperature}, "
                f"rain={self.rain_level}, events={self.adverse_events})")

class Forecast:
    """
    Wraps a list of ForecastEntry objects.

    Note the two attributes calculated in __init__. We work them out ONCE,
    when the data is loaded, instead of recalculating them every time a plane
    asks about the weather. Keep this in mind for the complexity question.
    """

    def __init__(self, entries):
        self.entries = entries                              # list[ForecastEntry]

        # Total projected rainfall = the sum of every entry's rain_level.
        self.cumulative_rain = sum(e.rain_level for e in entries)

        # IMPLEMENT RULE 2 HERE 
        
        ...

        self.has_heavy_weather = ...

    def __len__(self):
        return len(self.entries)

    def __repr__(self):
        return (f"Forecast({len(self.entries)} entries, "
                f"cumulative_rain={self.cumulative_rain:.1f}, "
                f"heavy_weather={self.has_heavy_weather})")
```

```python
class Airport:

    def __init__(self, location, city, max_capacity, current_occupancy, forecast):
        self.location = location                # str, e.g. "KBOS"
        self.city = city                        # str, e.g. "Boston"
        self.max_capacity = max_capacity        # int, total parking spots
        self.current_occupancy = current_occupancy   # int, spots currently USED
        self.forecast = forecast                # Forecast object

        # IMPLEMENT RULE 3 HERE
        self.is_high_capacity = ...


    def __repr__(self):
        return (f"Airport({self.location}, {self.city}, "
                f"{self.current_occupancy}/{self.max_capacity}, "
                f"{self.forecast!r})")

class Flight:
    # Lower number = higher priority. Medivac departs before VIP, and so on.
    PRIORITY_ORDER = {
        "Medivac": 0,
        "VIP": 1,
        "Commercial": 2,
        "Light Aircraft": 3,
    }

    def __init__(self, flight_id, designation, destination, max_rain_thresh):
        self.flight_id = flight_id              # str, e.g. "NMED001"
        self.designation = designation          # str, one of PRIORITY_ORDER
        self.destination = destination          # str, an Airport's location code
        self.max_rain_thresh = max_rain_thresh  # float, max rain it can land in
        self.priority = Flight.PRIORITY_ORDER[self.designation]

    def get_priority(self):
        return self.priority

    def __repr__(self):
        return (f"Flight({self.flight_id}, {self.designation}, "
                f"-> {self.destination})")


```

```python
def load_airports(path="airports.json"):
    """Read airports.json and return a list of Airport objects."""
    with open(path) as f:
        raw_airports = json.load(f)

    airports = []
    for raw in raw_airports:
        # Build a ForecastEntry for each dictionary in the "forecast" list...
        entries = [
            ForecastEntry(
                temperature=e["temperature"],
                rain_level=e["rain_level"],
                adverse_events=e["adverse_events"],
            )
            for e in raw["forecast"]
        ]
        # ...then wrap them all in one Forecast object.
        forecast = Forecast(entries)

        airports.append(Airport(
            location=raw["location"],
            city=raw["city"],
            max_capacity=raw["max_capacity"],
            current_occupancy=raw["current_occupancy"],
            forecast=forecast,
        ))
    return airports


def load_flights(path="flights.json"):
    """Read flights.json and return a list of Flight objects."""
    with open(path) as f:
        raw_flights = json.load(f)

    return [
        Flight(
            flight_id=raw["flight_id"],
            designation=raw["designation"],
            destination=raw["destination"],
            max_rain_thresh=raw["max_rain_thresh"],
        )
        for raw in raw_flights
    ]

def build_airport_lookup(airports):
    """
    Build a dictionary that maps a location code -> Airport object.
    We do this for quick O(1) access to different airports.
    """
    return {airport.location: airport for airport in airports}
```
