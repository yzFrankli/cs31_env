## **7. Mitigating Harms**

Now that you've identified and prioritized harms, take your top three harms and classify each (as best you can) as one of:

- **Data Privacy Harms:** 
    - What is it: Sensitive data is data that, if misused or made public, could cause harm to individuals (e.g. financial information, private medical information). 
    - Common harms: data leaks, sensitive data use, privacy violation through improper sharing of data. 
    - Possible Mitigations: follow appropriate data security procedures, reduce access to sensitive data, de-identify your data, remove sensitive features altogether, or don't collect sensitive data in the first place.
- **Model Security Harms:** 
    - What is it: Harms that arise from concerns around use, manipulation, or attacks on the model. 
    - Common harms: intentional deception in input data, corrupted or manipulated training data, exploiting the structure of a model to control decisions. 
    - Possible Mitigations: ensure data is not manipulated, don't train on live data, build systems that don't allow bad data, have human verification for some decisions, build protection to avoid users probing the model to learn its structure.
- **Accountability Harms:** 
    - What is it: A model, as an AI model, cannot be held accountable -- so who is going to be accountable for the decision the model makes? 
    - Common harms: incorrect model behavior (false positives, false negatives), harmful and unexplained decisions, user confusion over outcomes. 
    - Possible Mitigations: introduce a human review process for a subset of the model decisions, use explainable models so users understand why a decision is made, allow the user to appeal the decision.
